# Basic tensor operations

````python
import torch
````

## Tensor creation and gradient operations

````python
# Defining random variables

x = torch.tensor(3.0)
b = torch.tensor(2.0, requires_grad=True)
w = torch.tensor(1.0, requires_grad=True)
````

````python
# Defining the output function
y = w * x + b
````

````python
# Compute derivatives on y 
y.backward()
````

dy/dw is the gradient of y with respect to w, which is a tensor stored in w.grad

Same things for all variables where requires_grad=True

```python

````python
# Displaying gradients
print("dy/db:", b.grad) # Gradient of y with respect to b 
print("dy/dw:", w.grad) # Gradient of y with respect to w
print("dy/dx:", x.grad) # Gradient of y with respect to x (should be None since x does not require grad)

# This returns a tensor. If you want the value as a float, you can use .item()
print("dy/db (as float):", b.grad.item())  # Gradient of y with respect to b as float
````

## Tensor interaction with numpy

````python
import numpy as np
````

````python
r = np.random.rand(3, 2)

# Convert to a PyTorch tensor
t = torch.tensor(r, dtype=torch.float32)
````

````python
# Get the type of each
print("Type of tensor t:", t.dtype)  # Should print torch.float32
print("Type of numpy array r:", r.dtype)  # Should print float64
````

````python
# Convert from PyTorch tensor to numpy array
r_converted = t.numpy()
````
